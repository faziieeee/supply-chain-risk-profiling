```python
# ============================================================
# Cell 2: Define Risk Factors, Weights & Supplier Data
# ============================================================

# Risk factors and their weights (must sum to 1.0)
weights = {
    'Vendor_Financial_Health':      0.20,
    'Third_Party_Access_Control':   0.20,
    'Incident_Response_Readiness':  0.15,
    'Vulnerability_Management':     0.20,
    'Delivery_SLA_Performance':     0.15,
    'Compliance_Certification':     0.10
}

# Supplier data - scores out of 10
data = {
    'Supplier': ['Supplier A', 'Supplier B', 'Supplier C', 'Supplier D', 'Supplier E'],
    'Vendor_Financial_Health':      [7, 4, 8, 3, 6],
    'Third_Party_Access_Control':   [6, 8, 5, 7, 4],
    'Incident_Response_Readiness':  [5, 3, 9, 4, 7],
    'Vulnerability_Management':     [8, 5, 7, 2, 6],
    'Delivery_SLA_Performance':     [6, 7, 8, 5, 3],
    'Compliance_Certification':     [9, 4, 7, 3, 8]
}

df = pd.DataFrame(data)
print("✅ Supplier Data Loaded!")
print(df)
```

```python
# ============================================================
# Cell 3: Calculate Weighted Risk Scores
# ============================================================

# List of factor columns
factors = list(weights.keys())

# Calculate weighted score for each supplier
# Formula: Risk Score = Sum(Weight_i x Score_i)
df['Risk_Score'] = sum(df[factor] * weights[factor] for factor in factors)

# Classify risk level based on score
def classify_risk(score):
    if score <= 3:
        return '🟢 Low'
    elif score <= 5.5:
        return '🟡 Medium'
    elif score <= 7.5:
        return '🟠 High'
    else:
        return '🔴 Critical'

df['Risk_Level'] = df['Risk_Score'].apply(classify_risk)

# Display results sorted by risk score
result = df[['Supplier', 'Risk_Score', 'Risk_Level']].sort_values('Risk_Score', ascending=False)
result = result.reset_index(drop=True)

print("✅ Risk Scores Calculated!")
print("=" * 45)
print(result.to_string(index=False))
print("=" * 45)
print(f"\nHighest Risk: {result.iloc[0]['Supplier']} ({result.iloc[0]['Risk_Score']:.2f})")
print(f"Lowest Risk:  {result.iloc[-1]['Supplier']} ({result.iloc[-1]['Risk_Score']:.2f})")
```

```python
# ============================================================
# Cell 4: Visualize Risk Scores
# ============================================================

fig, axes = plt.subplots(1, 2, figsize=(14, 6))
fig.suptitle('Supply Chain Cyber Risk Profiling Dashboard', 
             fontsize=16, fontweight='bold', y=1.02)

# Colors based on risk level
colors = ['#FF6B6B' if s >= 7 else '#FFA500' if s >= 5.5 else '#4CAF50' 
          for s in result['Risk_Score']]

# Chart 1: Risk Score Bar Chart
axes[0].barh(result['Supplier'], result['Risk_Score'], color=colors)
axes[0].set_xlabel('Risk Score (0-10)')
axes[0].set_title('Supplier Risk Scores')
axes[0].axvline(x=5.5, color='orange', linestyle='--', label='High Risk Threshold')
axes[0].axvline(x=7.5, color='red', linestyle='--', label='Critical Threshold')
axes[0].legend()
axes[0].set_xlim(0, 10)

# Chart 2: Factor Weight Pie Chart
axes[1].pie(weights.values(), labels=weights.keys(), autopct='%1.0f%%',
            startangle=140, colors=sns.color_palette('Set2'))
axes[1].set_title('Risk Factor Weights')

plt.tight_layout()
plt.savefig('../reports/risk_dashboard.png', dpi=150, bbox_inches='tight')
plt.show()
print("✅ Charts saved to reports folder!")
```

```python
# ============================================================
# Cell 5: Export Results & Final Summary
# ============================================================

# Add individual weighted scores for each factor
for factor in factors:
    df[f'Weighted_{factor}'] = df[factor] * weights[factor]

# Export to CSV
df.to_csv('../reports/risk_profiling_results.csv', index=False)

# Final Summary
print("=" * 55)
print("   SUPPLY CHAIN CYBER RISK PROFILING — FINAL REPORT")
print("=" * 55)
print(f"\n📊 Total Suppliers Assessed : {len(df)}")
print(f"📌 Risk Factors Evaluated  : {len(factors)}")
print(f"\n🔴 High Risk Suppliers     : {len(df[df['Risk_Score'] >= 5.5])}")
print(f"🟡 Medium Risk Suppliers   : {len(df[(df['Risk_Score'] >= 3) & (df['Risk_Score'] < 5.5)])}")
print(f"🟢 Low Risk Suppliers      : {len(df[df['Risk_Score'] < 3])}")
print(f"\n📈 Average Risk Score      : {df['Risk_Score'].mean():.2f}")
print(f"⚠️  Highest Risk Supplier  : {df.loc[df['Risk_Score'].idxmax(), 'Supplier']}")
print(f"✅ Lowest Risk Supplier    : {df.loc[df['Risk_Score'].idxmin(), 'Supplier']}")
print("\n📁 Results exported to: reports/risk_profiling_results.csv")
print("📁 Dashboard saved to : reports/risk_dashboard.png")
print("\n✅ Project PRJN26-250 Complete!")
print("=" * 55)
```
